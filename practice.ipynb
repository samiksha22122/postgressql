Here's how you can **create a PostgreSQL database, user, and password** using the PostgreSQL shell (`psql`) or terminal commands.

---

## 🛠️ Step-by-Step Guide

### ✅ Step 1: Open the PostgreSQL Shell

If you're on Linux/Mac:

```bash
sudo -u postgres psql
```

On Windows:

* Open **pgAdmin** or use **SQL Shell (psql)** from Start menu.
* Login as superuser (`postgres`), or enter as prompted.

---

### ✅ Step 2: Create a New User

```sql
CREATE USER myuser WITH PASSWORD 'mypassword';
```

---

### ✅ Step 3: Create a New Database

```sql
CREATE DATABASE mydatabase;
```

---

### ✅ Step 4: Grant Privileges

```sql
GRANT ALL PRIVILEGES ON DATABASE mydatabase TO myuser;
```

---

### ✅ Step 5: Exit

```sql
\q
```

---

## ✅ Now Use It in Python

```python
import psycopg2

conn = psycopg2.connect(
    dbname="mydatabase",
    user="myuser",
    password="mypassword",
    host="localhost",
    port="5432"
)
```

---

## 🧪 Tip: Verify with SQL

You can verify your users and databases:

```sql
\du      -- shows users
\l       -- shows databases
```

TABLE CREATION

````python
import psycopg2

conn = psycopg2.connect(
    dbname="ipro_db",
    user="supplychain",
    password="supplychain",
    host="localhost",
    port="5432"
)
cur = conn.cursor()

cur.execute("""
    CREATE TABLE IF NOT EXISTS test_table (
        id SERIAL PRIMARY KEY,
        name TEXT NOT NULL
    );
""")
conn.commit()
cur.close()
conn.close()
print("Connected and table created successfully!")
````

INSERTION

````python
# Connect to your PostgreSQL database
conn = psycopg2.connect(
    dbname="ipro_db",
    user="supplychain",
    password="supplychain",
    host="localhost",
    port="5432"
)

# Create a cursor object
cur = conn.cursor()

# Insert data
cur.execute("""
    INSERT INTO test_table (name)
    VALUES (%s), (%s), (%s);
""", ('Apple', 'Banana', 'Cherry'))

# Commit the changes
conn.commit()

# Close the connection
cur.close()
conn.close()

print("Data inserted successfully!")
````

DATA FETCHING

````python
# Reconnect
conn = psycopg2.connect(
    dbname="ipro_db",
    user="supplychain",
    password="supplychain",
    host="localhost",
    port="5432"
)

# Create a cursor
cur = conn.cursor()

# Fetch all data
cur.execute("SELECT * FROM test_table;")
rows = cur.fetchall()

# Print rows
for row in rows:
    print(f"ID: {row[0]}, Name: {row[1]}")
# Close
cur.close()
conn.close()
````

CREATING A CLASS FOR ALL THE STEPS

````python
class DBManager:
    def __init__(self):
        self.conn = psycopg2.connect(
            dbname="ipro_db",
            user="supplychain",
            password="supplychain",
            host="localhost",
            port="5432"
        )
        self.cur = self.conn.cursor()

    def insert(self, name):
        self.cur.execute("INSERT INTO test_table (name) VALUES (%s);", (name,))
        self.conn.commit()

    def fetch(self):
        self.cur.execute("SELECT * FROM test_table;")
        return self.cur.fetchall()

    def close(self):
        self.cur.close()
        self.conn.close()

# Usage
db = DBManager()
db.insert('Grapes')
for row in db.fetch():
    print(row)
db.close()
````
